## 2. Variabel dan Tipe Data Dasar

Python mendukung berbagai tipe data: `str`, `int`, `float`, dan `bool`. Variabel dibuat tanpa deklarasi tipe secara eksplisit.

```python
nama     = "Mahasiswa Informatika"  # String
angkatan = 2024                     # Integer
ipk      = 3.75                     # Float
aktif    = True                     # Boolean

print(f"Nama     : {nama}")
print(f"Angkatan : {angkatan}")
print(f"IPK      : {ipk}")
print(f"Aktif    : {aktif}")
```

## 3. List dan Perulangan

Struktur data `list` menyimpan kumpulan elemen. Perulangan `for` dengan `enumerate()` memungkinkan iterasi sekaligus mendapatkan indeks.

```python
tools = ["Python", "Google Colab", "Jupyter Notebook", "GitHub"]
print("Tools Data Science yang akan kita pelajari:")
for i, tool in enumerate(tools, 1):
    print(f"  {i}. {tool}")
```

## 4. Fungsi Sederhana

Fungsi (`def`) mengemas logika yang dapat dipanggil kembali. Docstring digunakan untuk mendokumentasikan fungsi.

```python
def perkenalan(nama, prodi, angkatan):
    """Fungsi untuk menampilkan perkenalan mahasiswa."""
    return f"Hai! Saya {nama} dari {prodi}, angkatan {angkatan}."

hasil = perkenalan("Budi Santoso", "Informatika", 2024)
print(hasil)
```
