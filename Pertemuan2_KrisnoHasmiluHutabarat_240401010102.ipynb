```python
# ── Q1: Struktur Dataset Titanic ──

print("Jumlah baris  :", df.shape[0])
print("Jumlah kolom  :", df.shape[1])
print("\nNama kolom:")
print(df.columns.tolist())
```

## Q1 — Struktur Dataset Titanic
Dataset Titanic memiliki **891 baris** dan **12 kolom**.
Kolom yang tersedia: PassengerId, Survived, Pclass, Name, Sex, Age, SibSp, Parch, Ticket, Fare, Cabin, Embarked.

```python
# ── Q2: Missing Values ──

missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Jumlah Missing': missing_count,
    'Proporsi (%)': missing_pct
})

# Tampilkan hanya kolom yang ada missing values
print(missing_df[missing_df['Jumlah Missing'] > 0])
```

## Q2 — Missing Values
Terdapat 3 kolom dengan missing values:
- **Cabin**: 687 missing (77.10%) — sangat banyak
- **Age**: 177 missing (19.87%) — perlu diperhatikan
- **Embarked**: 2 missing (0.22%) — dapat diabaikan

```python
# ── Q3: Persentase Penumpang Selamat ──

total = len(df)
selamat = df['Survived'].sum()
pct_selamat = (selamat / total * 100).round(2)

print("Total penumpang :", total)
print("Jumlah selamat  :", selamat)
print("Tidak selamat   :", total - selamat)
print(f"Persentase selamat: {pct_selamat}%")
```

## Q3 — Persentase Penumpang Selamat
Hanya **38.38%** (342 dari 891) penumpang yang selamat.
Sebanyak 549 penumpang (61.62%) tidak selamat.
Hal ini sesuai fakta sejarah karena jumlah sekoci Titanic
tidak mencukupi untuk semua penumpang.

```python
# ── Q4: Penumpang Wanita Kelas 1 ──

wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

print("Jumlah penumpang wanita kelas 1:", len(wanita_kelas1))
print("Rata-rata usia                 :", wanita_kelas1['Age'].mean().round(2))
```

## Q4 — Penumpang Wanita Kelas 1
Terdapat **94 penumpang wanita** di kelas 1 dengan rata-rata usia **34.61 tahun**.
Wanita kelas 1 merupakan kalangan atas yang kemungkinan besar
mendapat prioritas evakuasi ("women and children first").

```python
# ── Q5: Tingkat Keselamatan per Kelas ──

survival_per_kelas = df.groupby('Pclass')['Survived'].mean().round(4) * 100

print("Tingkat keselamatan per kelas (%):")
print(survival_per_kelas)
print("\nKelas dengan tingkat keselamatan tertinggi: Kelas", survival_per_kelas.idxmax())
```

## Q5 — Tingkat Keselamatan per Kelas
Kelas 1: 62.96% | Kelas 2: 47.28% | Kelas 3: 24.24%
**Kelas 1 memiliki tingkat keselamatan tertinggi.**
Penumpang kelas atas mendapat akses lebih mudah ke sekoci
dibanding penumpang kelas bawah.

```python
# ── Q6: Tingkat Keselamatan Anak-anak vs Dewasa ──

# Hapus data yang Age-nya kosong
df_age = df.dropna(subset=['Age'])

# Kategorikan anak-anak dan dewasa
df_age = df_age.copy()
df_age['Kategori'] = df_age['Age'].apply(lambda x: 'Anak-anak (<18)' if x < 18 else 'Dewasa (>=18)')

# Hitung tingkat keselamatan
hasil = df_age.groupby('Kategori')['Survived'].agg(
    Jumlah='count',
    Selamat='sum',
).reset_index()
hasil['Pct_Selamat (%)'] = (hasil['Selamat'] / hasil['Jumlah'] * 100).round(2)

print("Tingkat Keselamatan Anak-anak vs Dewasa:")
print(hasil.to_string(index=False))
```

## Q6 — Anak-anak vs Dewasa
Anak-anak (<18 tahun): **53.98%** selamat
Dewasa (>=18 tahun): **38.10%** selamat

Anak-anak memiliki tingkat keselamatan 15% lebih tinggi dari dewasa.
Hal ini membuktikan kebijakan evakuasi "women and children first"
benar-benar diterapkan saat tragedi Titanic tenggelam.
